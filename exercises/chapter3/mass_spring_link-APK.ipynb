```python
import numpy as np
import cvxpy as cp
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
```

```python
# Problem data.
m = .02 # mass
g = 10 # gravity acceleration
k = 100 # stiffness
n = 100 # number of mass points
eps = 1e-4
# Initial state.
q0 = np.array([0,1])
qn = np.array([n-1,1])
D = -1*np.eye(n) + np.eye(n, k=1)
D = D[:-1,:]
```

```python
# Variables.
Q = cp.Variable((n, 2)) 

# Initialize empty list of constraints.
constraints = []

constraints.append(Q[0] == q0)
constraints.append(Q[-1] == qn)
constraints.append(Q[:, 1] >= 0)

# Objective function.
# obj = m*g*np.ones((n,1)).T @ Q @ np.array([0,1]).T + 0.5*k*cp.trace(D@Q@Q.T@D.T)
obj = m*g*cp.sum(Q[:,1]) + 0.5*k*cp.sum_squares(Q[1:] - Q[:-1])
# obj = m*g*np.ones((n,1)).T @ Q @ np.array([0,1]).T + 0.5*k*cp.sum_squares(cp.diff(Q[1:,:] - Q[:-1,:], axis=1))
Q.value = np.linspace(q0, qn, n)
```

```python
# Solve problem.
prob = cp.Problem(cp.Minimize(obj), constraints)
prob.solve(nlp=True, solver=cp.IPOPT)
```
